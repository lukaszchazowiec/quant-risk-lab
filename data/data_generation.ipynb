```python
import os
import numpy as np
import pandas as pd
import yfinance as yf
```

```python
tickers = [
    "SPY",      # S&P 500 (Market Benchmark)
    "IWM",      # Russell 2000 (Small-Cap Benchmark)
    "NVDA",     # Tech / High Growth / High Vol
    "TSLA",     # High Beta / Volatility Clustering
    "JPM",      # Financials / Interest Rate Sensitivity
    "XOM",      # Energy / Commodity Sensitivity
    "PG",       # Consumer Staples / Low Vol Defensive
    "JNJ",      # Healthcare / Low Vol Defensive
    "GLD",      # Gold / Inflation & Market Hedge
    "BTC-USD"   # Crypto / Extreme Volatility & Heavy Tails
]
```

```python
folder = r"D:\LukaszPliki\FinanceProjectsPython\quant-risk-lab\data"
os.makedirs(folder, exist_ok=True)

prices = yf.download(tickers, start="2000-01-01", auto_adjust=True)["Close"]
prices.to_csv(folder + r"\prices.csv")

print(prices.shape)
```
