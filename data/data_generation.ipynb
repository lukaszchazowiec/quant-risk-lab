```python
from pathlib import Path
import pandas as pd
import yfinance as yf
```

```python
tickers = [
    "SPY",      # S&P 500 (Market Benchmark)
    "IWM",      # Russell 2000 (Small-Cap Benchmark)
    "NVDA",     # Tech / High Growth / High Vol
    "AMZN",     # Growth / Dot-com crash and 2008 drawdowns
    "JPM",      # Financials / Interest Rate Sensitivity
    "XOM",      # Energy / Commodity Sensitivity
    "PG",       # Consumer Staples / Low Vol Defensive
    "JNJ",      # Healthcare / Low Vol Defensive
    "GC=F",     # Gold Futures / Inflation & Market Hedge
    "VUSTX"     # Long-Term Treasuries / Flight-to-Quality Hedge
]
```

```python
DATA = Path("data")
DATA.mkdir(exist_ok=True)
```

```python
prices = yf.download(tickers, start="2000-09-01", auto_adjust=True)["Close"]
prices.to_csv(DATA / "prices.csv")
print(prices.shape)
```
